```python
import pandas as pd
df = pd.read_csv("/kaggle/input/datasets/organizations/uciml/pima-indians-diabetes-database/diabetes.csv")
print(df.describe())
print(df.isnull().sum())
print("First 5 rows:")
print(df.head())
print("Last 5 rows:")
print(df.tail())
print("number of rows and columns :" ,df.shape)
print(df.columns)
print(df.dtypes)
print("average age :" , df["Age"].mean())
print("average bmi :" , df["BMI"].mean())
print("highest glucose :" ,df["Glucose"].max())
print("lowest glucose :" ,df["Glucose"].min())
print("diabetic patient :" ,(df["Outcome"] == 1).sum())
print("non-diabetic patient :" ,(df["Outcome"] == 0).sum())
print(df.groupby("Outcome")["BMI"].mean())
print(df.groupby("Outcome")["Glucose"].mean())
print(df.groupby("Outcome")["Age"].mean())
```
