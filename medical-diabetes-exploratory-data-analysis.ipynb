```python
# Medical Diabetes Exploratory Data Analysis
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/kaggle/input/datasets/organizations/uciml/pima-indians-diabetes-database/diabetes.csv",index_col = 0)
bins_num = [20,30,40,50,60,70,80]
age = plt.hist(df["Age"],bins = bins_num , edgecolor='black')
plt.xlabel("age in years")
plt.ylabel("number of patient")
plt.title("Age Distribution")
plt.show()
bins_num = [18.5,25,30,35,40]
bmi = plt.hist(df["BMI"],bins = bins_num , edgecolor='black')
plt.xlabel("bmi")
plt.ylabel("number of patient")
plt.title("BMI OF DIABETIC PATIENT")
plt.show()
bins_glucose = [0, 70, 100, 125, 160, 200]
glucose = plt.hist(df["Glucose"],bins = bins_glucose , edgecolor='black')
plt.xlabel("glucose value")
plt.ylabel("number of patient")
plt.title("GLUCOSE OF DIABETIC PATIENT")
plt.show()
relation = plt.scatter(df.BMI,df.Glucose)
plt.xlabel("BMI")
plt.ylabel("Glucose Value")
plt.title("RELATION BTW BMI AND GLUCOSE")
plt.show()
outcome_count = df["Outcome"].value_counts()
print(outcome_count)
result = plt .bar(outcome_count.index,outcome_count.values, edgecolor='black')
plt.xlabel("Outcome (0 = NON-DIABETIC ,1 = DIABETIC)")
plt.ylabel("Count")
plt.title("Distribution of Diabetic vs Non-Diabetic Patients")
plt.xticks([0,1],["Non-Diabetic (0)", "Diabetic (1)"])
plt.show()
```

# Medical Diabetes Exploratory Data Analysis

## Introduction

Diabetes Mellitus is a chronic medical condition that significantly impacts global public health. Early detection and data-driven analysis play a critical role in evaluating risk factors. This Exploratory Data Analysis (EDA) investigates the relationship between demographic characteristics, clinical parameters, and diabetes diagnosis to extract meaningful medical insights.

## Dataset Information

This project utilizes the Pima Indians Diabetes Database. The dataset includes several clinical measurements collected from female patients of Pima Indian heritage aged 21 and older.Glucose: Plasma glucose concentration (2 hours in an oral glucose tolerance test).BMI: Body Mass Index ($\text{weight in kg} / (\text{height in m})^2$).Age: Age in years.Outcome: Binary class variable ($0 = \text{Non-Diabetic}$, $1 = \text{Diabetic}$).

## Age Distribution

Observation: The majority of patients in this dataset are young adults aged between 20 and 30 years old.

## BMI Distribution

Observation: Most patients fall into the overweight ($\text{BMI } 25\text{--}30$) and obese ($\text{BMI } > 30$) categories, indicating a high prevalence of elevated BMI in the study population.

## Glucose Distribution

Observation: A significant proportion of patients exhibit glucose levels above normal ranges ($>100 \text{ mg/dL}$), reflecting elevated risk or pre-existing diabetic conditions

## Relationship Between BMI and Glucose

Observation: There is a positive trend between BMI and blood glucose levels, suggesting that higher body mass index correlates with increased blood sugar levels.

## Outcome Distribution

Observation: There is a positive trend between BMI and blood glucose levels, suggesting that higher body mass index correlates with increased blood sugar levels.

## Conclusions

1.Age Factor: The dataset primarily consists of younger individuals, with the highest concentration of patients under 30 years old.

2.High Prevalence of Obesity: The vast majority of individuals have a BMI >= 25, establishing obesity as a prominent characteristic within this cohort.

3.Elevated Blood Glucose: A notable percentage of individuals exhibit elevated blood glucose values (>=100 mg/dL), aligning with key diagnostic indicators for diabetes.

4.Positive BMI-Glucose Correlation: Higher BMI values tend to associate with higher blood glucose levels, emphasizing metabolic risk associated with weight.

5.Class Imbalance: The target outcome is imbalanced (500 non-diabetic vs. 268 diabetic), which is a critical consideration for future machine learning model training and evaluation metrics.
